## 4. Training checklist in Applio

1. Select the reviewed dataset folder.
2. Create a new experiment name: `AkTextToVoice_MyVoice`.
3. Choose a compatible architecture, sample rate, embedder and matching pretrained G/D pair.
4. Preprocess the dataset.
5. Extract pitch and content features.
6. Start with a conservative batch size that fits the available GPU memory.
7. Train, export the inference `.pth`, and generate the matching `.index` if supported.
8. Test with a held-out recording that was not included in training.

Keep the exported model and matching index together. Do not upload private voice recordings or model files to a public repository without deciding to share them.
